```python
import sys
import os
sys.path.insert(0, os.getcwd())

# Import config first
import config

# Now import data loader
from data_loader import load_and_validate_data, split_data

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import (classification_report, confusion_matrix, roc_auc_score, 
                             roc_curve, precision_recall_curve, f1_score, accuracy_score)
from imblearn.over_sampling import SMOTE
import pickle

print("✅ All imports successful")
print(f"Dataset path: {config.DATA_PATH}")
print(f"Models will be saved to: {config.MODELS_DIR}")
```

```python
# ============================================================================
# CELL 2: LOAD AND VALIDATE DATA
# ============================================================================

print("\n" + "="*80)
print("STEP 1: LOAD AND VALIDATE DATA")
print("="*80)

# Load the raw data
df = load_and_validate_data(DATA_PATH)
print(f"\n✅ Dataset loaded successfully: {df.shape}")

# Display first few rows
print("\nFirst 5 rows:")
print(df.head())

print("\nDataset Info:")
print(df.info())

print("\nBasic Statistics:")
print(df.describe())
```

```python
# ============================================================================
# CELL 3: SPLIT DATA INTO TRAIN/VAL/TEST
# ============================================================================

print("\n" + "="*80)
print("STEP 2: STRATIFIED TRAIN/VALIDATION/TEST SPLIT (70/15/15)")
print("="*80)

# Split data
train_df, val_df, test_df = split_data(df)

# Extract features and target
X_train = train_df.drop(columns=[CLEAN_TARGET])
y_train = train_df[CLEAN_TARGET]

X_val = val_df.drop(columns=[CLEAN_TARGET])
y_val = val_df[CLEAN_TARGET]

X_test = test_df.drop(columns=[CLEAN_TARGET])
y_test = test_df[CLEAN_TARGET]

print(f"\nTrain set: {X_train.shape}")
print(f"Val set: {X_val.shape}")
print(f"Test set: {X_test.shape}")

print(f"\n✅ Data split complete. Ready for feature engineering.")
```

```python
# ============================================================================
# CELL 4: FEATURE ENGINEERING & PREPROCESSING
# ============================================================================

print("\n" + "="*80)
print("STEP 3: FEATURE ENGINEERING & PREPROCESSING")
print("="*80)

from sklearn.preprocessing import LabelEncoder, StandardScaler
from imblearn.over_sampling import SMOTE

# Step 1: Encode categorical features
print("\n[1] Encoding categorical features...")
le = LabelEncoder()
X_train_encoded = X_train.copy()
X_val_encoded = X_val.copy()
X_test_encoded = X_test.copy()

# Fit encoder on train set only (no data leakage)
X_train_encoded[CLEAN_CATEGORICAL_FEATURE] = le.fit_transform(X_train[CLEAN_CATEGORICAL_FEATURE])
X_val_encoded[CLEAN_CATEGORICAL_FEATURE] = le.transform(X_val[CLEAN_CATEGORICAL_FEATURE])
X_test_encoded[CLEAN_CATEGORICAL_FEATURE] = le.transform(X_test[CLEAN_CATEGORICAL_FEATURE])

print(f"   Machine types encoded: {dict(zip(le.classes_, le.transform(le.classes_)))}")

# Step 2: Scale numeric features
print("\n[2] Scaling numeric features...")
scaler = StandardScaler()
numeric_cols = CLEAN_NUMERIC_FEATURES

# Fit scaler on train set only
X_train_scaled = X_train_encoded.copy()
X_train_scaled[numeric_cols] = scaler.fit_transform(X_train_encoded[numeric_cols])

X_val_scaled = X_val_encoded.copy()
X_val_scaled[numeric_cols] = scaler.transform(X_val_encoded[numeric_cols])

X_test_scaled = X_test_encoded.copy()
X_test_scaled[numeric_cols] = scaler.transform(X_test_encoded[numeric_cols])

print(f"   Numeric columns scaled: {numeric_cols}")

# Step 3: Apply SMOTE to balance training data
print("\n[3] Applying SMOTE to training data...")
smote = SMOTE(random_state=RANDOM_STATE)
X_train_smote, y_train_smote = smote.fit_resample(X_train_scaled, y_train)

print(f"   Before SMOTE: {y_train.value_counts().to_dict()}")
print(f"   After SMOTE:  {pd.Series(y_train_smote).value_counts().to_dict()}")
print(f"   New training set size: {X_train_smote.shape[0]} rows")

# Store processed data for next cells
print("\n✅ Feature preprocessing complete.")
print(f"   Final feature set: {list(X_train_smote.columns)}")
```

```python
# ============================================================================
# CELL 5: TRAIN ALL CANDIDATE MODELS (PROFESSIONAL COMPARISON)
# ============================================================================

print("\n" + "="*80)
print("STEP 4: TRAIN & EVALUATE ALL CANDIDATE MODELS")
print("="*80)

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, 
                             f1_score, roc_auc_score, confusion_matrix, classification_report)
import xgboost as xgb

# Dictionary to store models and results
models = {}
results = {}

# ===== MODEL 1: LOGISTIC REGRESSION =====
print("\n[1/5] Training Logistic Regression...")
lr = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE, class_weight='balanced')
lr.fit(X_train_smote, y_train_smote)
y_val_pred_lr = lr.predict(X_val_scaled)
y_val_prob_lr = lr.predict_proba(X_val_scaled)[:, 1]

models['Logistic Regression'] = lr
results['Logistic Regression'] = {
    'accuracy': accuracy_score(y_val, y_val_pred_lr),
    'precision': precision_score(y_val, y_val_pred_lr),
    'recall': recall_score(y_val, y_val_pred_lr),
    'f1': f1_score(y_val, y_val_pred_lr),
    'roc_auc': roc_auc_score(y_val, y_val_prob_lr)
}
print(f"   ✓ F1: {results['Logistic Regression']['f1']:.4f}, ROC-AUC: {results['Logistic Regression']['roc_auc']:.4f}")

# ===== MODEL 2: DECISION TREE =====
print("\n[2/5] Training Decision Tree...")
dt = DecisionTreeClassifier(max_depth=10, random_state=RANDOM_STATE, class_weight='balanced')
dt.fit(X_train_smote, y_train_smote)
y_val_pred_dt = dt.predict(X_val_scaled)
y_val_prob_dt = dt.predict_proba(X_val_scaled)[:, 1]

models['Decision Tree'] = dt
results['Decision Tree'] = {
    'accuracy': accuracy_score(y_val, y_val_pred_dt),
    'precision': precision_score(y_val, y_val_pred_dt),
    'recall': recall_score(y_val, y_val_pred_dt),
    'f1': f1_score(y_val, y_val_pred_dt),
    'roc_auc': roc_auc_score(y_val, y_val_prob_dt)
}
print(f"   ✓ F1: {results['Decision Tree']['f1']:.4f}, ROC-AUC: {results['Decision Tree']['roc_auc']:.4f}")

# ===== MODEL 3: RANDOM FOREST =====
print("\n[3/5] Training Random Forest...")
rf = RandomForestClassifier(n_estimators=100, max_depth=15, random_state=RANDOM_STATE, 
                            class_weight='balanced', n_jobs=-1)
rf.fit(X_train_smote, y_train_smote)
y_val_pred_rf = rf.predict(X_val_scaled)
y_val_prob_rf = rf.predict_proba(X_val_scaled)[:, 1]

models['Random Forest'] = rf
results['Random Forest'] = {
    'accuracy': accuracy_score(y_val, y_val_pred_rf),
    'precision': precision_score(y_val, y_val_pred_rf),
    'recall': recall_score(y_val, y_val_pred_rf),
    'f1': f1_score(y_val, y_val_pred_rf),
    'roc_auc': roc_auc_score(y_val, y_val_prob_rf)
}
print(f"   ✓ F1: {results['Random Forest']['f1']:.4f}, ROC-AUC: {results['Random Forest']['roc_auc']:.4f}")

# ===== MODEL 4: XGBOOST =====
print("\n[4/5] Training XGBoost...")
xgb_model = xgb.XGBClassifier(n_estimators=100, max_depth=7, learning_rate=0.1, 
                               random_state=RANDOM_STATE, scale_pos_weight=len(y_train_smote[y_train_smote==0])/len(y_train_smote[y_train_smote==1]),
                               use_label_encoder=False, eval_metric='logloss')
xgb_model.fit(X_train_smote, y_train_smote)
y_val_pred_xgb = xgb_model.predict(X_val_scaled)
y_val_prob_xgb = xgb_model.predict_proba(X_val_scaled)[:, 1]

models['XGBoost'] = xgb_model
results['XGBoost'] = {
    'accuracy': accuracy_score(y_val, y_val_pred_xgb),
    'precision': precision_score(y_val, y_val_pred_xgb),
    'recall': recall_score(y_val, y_val_pred_xgb),
    'f1': f1_score(y_val, y_val_pred_xgb),
    'roc_auc': roc_auc_score(y_val, y_val_prob_xgb)
}
print(f"   ✓ F1: {results['XGBoost']['f1']:.4f}, ROC-AUC: {results['XGBoost']['roc_auc']:.4f}")

# ===== MODEL 5: HISTOGRAM GRADIENT BOOSTING =====
print("\n[5/5] Training HistGradientBoosting...")
hgb = HistGradientBoostingClassifier(learning_rate=0.1, max_iter=100, random_state=RANDOM_STATE)
hgb.fit(X_train_smote, y_train_smote)
y_val_pred_hgb = hgb.predict(X_val_scaled)
y_val_prob_hgb = hgb.predict_proba(X_val_scaled)[:, 1]

models['HistGradientBoosting'] = hgb
results['HistGradientBoosting'] = {
    'accuracy': accuracy_score(y_val, y_val_pred_hgb),
    'precision': precision_score(y_val, y_val_pred_hgb),
    'recall': recall_score(y_val, y_val_pred_hgb),
    'f1': f1_score(y_val, y_val_pred_hgb),
    'roc_auc': roc_auc_score(y_val, y_val_prob_hgb)
}
print(f"   ✓ F1: {results['HistGradientBoosting']['f1']:.4f}, ROC-AUC: {results['HistGradientBoosting']['roc_auc']:.4f}")

# ===== PROFESSIONAL MODEL COMPARISON =====
print("\n" + "="*80)
print("MODEL COMPARISON ON VALIDATION SET (RANKED BY F1-SCORE)")
print("="*80)

results_df = pd.DataFrame(results).T
results_df = results_df.sort_values('f1', ascending=False)
print("\n", results_df.round(4))

# Identify top 2 models for tuning
top_2_models = results_df.head(2).index.tolist()
print(f"\n🏆 Top 2 Models for Hyperparameter Tuning:")
for i, model_name in enumerate(top_2_models, 1):
    f1 = results_df.loc[model_name, 'f1']
    auc = results_df.loc[model_name, 'roc_auc']
    print(f"   {i}. {model_name:25s} | F1: {f1:.4f} | ROC-AUC: {auc:.4f}")

print(f"\n✅ All 5 models trained successfully.")
```

```python
# ============================================================================
# CELL 6: HYPERPARAMETER TUNING (TOP 2 MODELS)
# ============================================================================

print("\n" + "="*80)
print("STEP 5: HYPERPARAMETER TUNING - TOP 2 MODELS")
print("="*80)

from sklearn.model_selection import GridSearchCV

tuned_models = {}
tuned_results = {}

# ===== TUNE MODEL 1: XGBOOST =====
print("\n[1/2] Tuning XGBoost...")
xgb_base = xgb.XGBClassifier(random_state=RANDOM_STATE, use_label_encoder=False, eval_metric='logloss')

xgb_param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [5, 7, 10],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.7, 0.9]
}

xgb_grid = GridSearchCV(xgb_base, xgb_param_grid, cv=5, scoring='f1', n_jobs=-1, verbose=0)
xgb_grid.fit(X_train_smote, y_train_smote)

tuned_models['XGBoost'] = xgb_grid.best_estimator_
y_val_pred_xgb_tuned = tuned_models['XGBoost'].predict(X_val_scaled)
y_val_prob_xgb_tuned = tuned_models['XGBoost'].predict_proba(X_val_scaled)[:, 1]

tuned_results['XGBoost'] = {
    'best_params': xgb_grid.best_params_,
    'cv_f1': xgb_grid.best_score_,
    'accuracy': accuracy_score(y_val, y_val_pred_xgb_tuned),
    'precision': precision_score(y_val, y_val_pred_xgb_tuned),
    'recall': recall_score(y_val, y_val_pred_xgb_tuned),
    'f1': f1_score(y_val, y_val_pred_xgb_tuned),
    'roc_auc': roc_auc_score(y_val, y_val_prob_xgb_tuned)
}

print(f"   Best params: {tuned_results['XGBoost']['best_params']}")
print(f"   CV F1-Score: {tuned_results['XGBoost']['cv_f1']:.4f}")
print(f"   Val F1-Score: {tuned_results['XGBoost']['f1']:.4f}")

# ===== TUNE MODEL 2: HISTGRADIENTBOOSTING =====
print("\n[2/2] Tuning HistGradientBoosting...")
hgb_base = HistGradientBoostingClassifier(random_state=RANDOM_STATE)

hgb_param_grid = {
    'learning_rate': [0.01, 0.05, 0.1],
    'max_iter': [100, 200, 300],
    'max_leaf_nodes': [15, 31, 63],
    'min_samples_leaf': [10, 20]
}

hgb_grid = GridSearchCV(hgb_base, hgb_param_grid, cv=5, scoring='f1', n_jobs=-1, verbose=0)
hgb_grid.fit(X_train_smote, y_train_smote)

tuned_models['HistGradientBoosting'] = hgb_grid.best_estimator_
y_val_pred_hgb_tuned = tuned_models['HistGradientBoosting'].predict(X_val_scaled)
y_val_prob_hgb_tuned = tuned_models['HistGradientBoosting'].predict_proba(X_val_scaled)[:, 1]

tuned_results['HistGradientBoosting'] = {
    'best_params': hgb_grid.best_params_,
    'cv_f1': hgb_grid.best_score_,
    'accuracy': accuracy_score(y_val, y_val_pred_hgb_tuned),
    'precision': precision_score(y_val, y_val_pred_hgb_tuned),
    'recall': recall_score(y_val, y_val_pred_hgb_tuned),
    'f1': f1_score(y_val, y_val_pred_hgb_tuned),
    'roc_auc': roc_auc_score(y_val, y_val_prob_hgb_tuned)
}

print(f"   Best params: {tuned_results['HistGradientBoosting']['best_params']}")
print(f"   CV F1-Score: {tuned_results['HistGradientBoosting']['cv_f1']:.4f}")
print(f"   Val F1-Score: {tuned_results['HistGradientBoosting']['f1']:.4f}")

# ===== COMPARE TUNED MODELS =====
print("\n" + "="*80)
print("TUNED MODEL COMPARISON")
print("="*80)

tuned_comparison = pd.DataFrame({
    model: {
        'CV F1': tuned_results[model]['cv_f1'],
        'Val Accuracy': tuned_results[model]['accuracy'],
        'Val Precision': tuned_results[model]['precision'],
        'Val Recall': tuned_results[model]['recall'],
        'Val F1': tuned_results[model]['f1'],
        'Val ROC-AUC': tuned_results[model]['roc_auc']
    }
    for model in tuned_results
}).T

print("\n", tuned_comparison.round(4))

best_final_model_name = tuned_comparison['Val F1'].idxmax()
print(f"\n🏆 FINAL CHAMPION: {best_final_model_name}")
print(f"   Validation F1-Score: {tuned_comparison.loc[best_final_model_name, 'Val F1']:.4f}")
print(f"   Validation ROC-AUC: {tuned_comparison.loc[best_final_model_name, 'Val ROC-AUC']:.4f}")

# Store the final model for next steps
best_final_model = tuned_models[best_final_model_name]
print(f"\n✅ Tuning complete. Ready for threshold optimization and test evaluation.")
```

```python
# ============================================================================
# CELL 7: THRESHOLD OPTIMIZATION & TEST SET EVALUATION
# ============================================================================

print("\n" + "="*80)
print("STEP 6: THRESHOLD OPTIMIZATION & FINAL TEST EVALUATION")
print("="*80)

from sklearn.metrics import precision_recall_curve, roc_curve

# Get probabilities on validation set
y_val_prob = best_final_model.predict_proba(X_val_scaled)[:, 1]

# ===== FIND OPTIMAL THRESHOLD =====
print("\n[1] Finding optimal threshold on validation set...")

# Calculate precision-recall curve
precision_curve, recall_curve, thresholds_pr = precision_recall_curve(y_val, y_val_prob)
f1_scores = 2 * (precision_curve * recall_curve) / (precision_curve + recall_curve + 1e-10)

# Find threshold that maximizes F1
optimal_idx = np.argmax(f1_scores)
optimal_threshold = thresholds_pr[optimal_idx] if optimal_idx < len(thresholds_pr) else 0.5

print(f"   Default threshold: 0.5")
print(f"   Optimal threshold (max F1): {optimal_threshold:.4f}")
print(f"   F1 at optimal threshold: {f1_scores[optimal_idx]:.4f}")

# ===== EVALUATE ON VALIDATION SET WITH OPTIMAL THRESHOLD =====
print("\n[2] Validation Set Performance (optimal threshold)...")
y_val_pred_optimal = (y_val_prob >= optimal_threshold).astype(int)

print(f"\n   Accuracy:  {accuracy_score(y_val, y_val_pred_optimal):.4f}")
print(f"   Precision: {precision_score(y_val, y_val_pred_optimal):.4f}")
print(f"   Recall:    {recall_score(y_val, y_val_pred_optimal):.4f}")
print(f"   F1-Score:  {f1_score(y_val, y_val_pred_optimal):.4f}")
print(f"   ROC-AUC:   {roc_auc_score(y_val, y_val_prob):.4f}")

print("\n   Confusion Matrix (Validation):")
cm_val = confusion_matrix(y_val, y_val_pred_optimal)
print(f"   {cm_val}")

# ===== EVALUATE ON TEST SET (FINAL EVALUATION) =====
print("\n" + "="*80)
print("[3] FINAL TEST SET EVALUATION")
print("="*80)

y_test_prob = best_final_model.predict_proba(X_test_scaled)[:, 1]
y_test_pred = (y_test_prob >= optimal_threshold).astype(int)

print(f"\n   Accuracy:  {accuracy_score(y_test, y_test_pred):.4f}")
print(f"   Precision: {precision_score(y_test, y_test_pred):.4f}")
print(f"   Recall:    {recall_score(y_test, y_test_pred):.4f}")
print(f"   F1-Score:  {f1_score(y_test, y_test_pred):.4f}")
print(f"   ROC-AUC:   {roc_auc_score(y_test, y_test_prob):.4f}")

print("\n   Confusion Matrix (Test):")
cm_test = confusion_matrix(y_test, y_test_pred)
print(f"   {cm_test}")
print(f"\n   TN: {cm_test[0,0]} | FP: {cm_test[0,1]}")
print(f"   FN: {cm_test[1,0]} | TP: {cm_test[1,1]}")

print("\n" + "="*80)
print("Classification Report (Test Set)")
print("="*80)
print(classification_report(y_test, y_test_pred, target_names=['No Failure', 'Failure']))

# ===== VISUALIZATIONS =====
print("\n[4] Generating visualizations...")

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_test_prob)
axes[0, 0].plot(fpr, tpr, label=f'XGBoost (AUC={roc_auc_score(y_test, y_test_prob):.4f})', linewidth=2)
axes[0, 0].plot([0, 1], [0, 1], 'k--', label='Random')
axes[0, 0].set_xlabel('False Positive Rate')
axes[0, 0].set_ylabel('True Positive Rate')
axes[0, 0].set_title('ROC Curve - Test Set')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Plot 2: Precision-Recall Curve
axes[0, 1].plot(recall_curve, precision_curve, label='XGBoost', linewidth=2)
axes[0, 1].set_xlabel('Recall')
axes[0, 1].set_ylabel('Precision')
axes[0, 1].set_title('Precision-Recall Curve - Validation Set')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# Plot 3: Confusion Matrix Heatmap
sns.heatmap(cm_test, annot=True, fmt='d', cmap='Blues', ax=axes[1, 0], 
            xticklabels=['No Failure', 'Failure'], yticklabels=['No Failure', 'Failure'])
axes[1, 0].set_title('Confusion Matrix - Test Set')
axes[1, 0].set_ylabel('True Label')
axes[1, 0].set_xlabel('Predicted Label')

# Plot 4: Feature Importance
feature_imp = pd.DataFrame({
    'Feature': X_test_scaled.columns,
    'Importance': best_final_model.feature_importances_
}).sort_values('Importance', ascending=True).tail(10)

axes[1, 1].barh(feature_imp['Feature'], feature_imp['Importance'], color='steelblue')
axes[1, 1].set_xlabel('Importance')
axes[1, 1].set_title('Top 10 Feature Importances (XGBoost)')

plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'model_evaluation.png'), dpi=300, bbox_inches='tight')
plt.show()

print(f"\n✅ Plots saved to: {PLOTS_DIR}")

# Store threshold for Streamlit app
OPTIMAL_THRESHOLD = optimal_threshold
print(f"\n✅ Optimal threshold stored: {OPTIMAL_THRESHOLD:.4f}")
```
